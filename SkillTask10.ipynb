```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
mp['hotel'] = le.fit_transform(mp['hotel'])

mp['meal'] = le.fit_transform(mp['meal'])

mp['country'] = le.fit_transform(mp['country'])

mp.head()
```

```python
sns.scatterplot(x='stays_in_weekend_nights', y='stays_in_week_nights', data=mp)

plt.title('Scatter Plot')
plt.xlabel('WEEKEND')
plt.ylabel('WEEK')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.violinplot(x='hotel',y='adr',data=mp,hue='babies',palette='Set2',legend=False)
plt.title('violin plot')
plt.xlabel('hotel')
plt.ylabel('adr')
plt.show()
```

```python
sns.boxplot(x='hotel', y='company', data=mp,color="blue")

plt.title('Box Plot')
plt.xlabel('Hotel')
plt.ylabel('company')
plt.show()
```

```python
print("Skewness:", mp['adr'].skew())
```

```python
sns.barplot(x='meal', y='babies', data=mp,color="pink")

plt.title('Bar Plot')
plt.xlabel('meal')
plt.ylabel('babies')
plt.show()
```
