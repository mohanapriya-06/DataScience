```python
sns.countplot(x='carrier', data=df)

plt.title(' FLIGHT')
plt.xlabel('Carrier')
plt.ylabel('Count')
plt.show()
```

```python
df['origin'].value_counts().plot.pie(autopct='%1.1f%%')

plt.title('Origin Airport')
plt.xlabel('origin')
plt.ylabel('Airport')
plt.show()
```

```python
df["origin"].value_counts().plot(kind="bar")

plt.title("barchart for Origin")
plt.xlabel("Origin")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x='origin', hue='carrier', data=df)

plt.title('Origin vs Carrier')
plt.show()
```
