```python
df=df.dropna()
```

```python
df.isnull()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
priya=df.select_dtypes(include=np.number).columns
print(priya)
```

```python
for column in priya:
    df[column]=df[column].fillna(df[column].mean())
```

```python
mp=df.select_dtypes(include="str").columns
print(mp)

for column in mp:
    df[column] = df[column].fillna(df[column].mode()[0])
```
