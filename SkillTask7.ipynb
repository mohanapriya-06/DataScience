```python
numeric_df=df.select_dtypes(include="number")
result=pd.DataFrame({
    "Mean":numeric_df.mean(),
    "Median":numeric_df.median(),
    "Mode":numeric_df.mode().iloc[0]
})
print(result.round(3))
```

```python
Q1=numeric_df.quantile(0.25)
Q3=numeric_df.quantile(0.75)

IQR=Q3-Q1
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR

outliers=(
    (numeric_df<lower_limit)|
    (numeric_df>upper_limit)
)
print("Lower Bound:\n",lower_limit)
print("Upper Bound:\n",upper_limit)
print("outliers in each column:\n",numeric_df[outliers].count())
```

```python
mean_value = numeric_df.mean()
median_value = numeric_df.median()

print("Mean:", mean_value)
print("Median:", median_value)

print("Skewness:")

for col in numeric_df.columns:
    if mean_value[col] > median_value[col]:
        print(col, "Positive Skewness")
    elif mean_value[col] < median_value[col]:
        print(col, "Negative Skewness")
    else:
        print(col, "Approximately Symmetrical")
```
