```python
plt.figure(figsize=(5,3))
plt.hist(mp['lead_time'], bins=10, color='blue', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Lead Time")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(x=mp['lead_time'])
plt.title("Box Plot")
plt.show()
```

```python
skeww = pd.DataFrame({
    "Skewness": mp.skew(numeric_only=True)
})

print(skeww.round(3))
```

```python
numeric_df = mp.select_dtypes(include=['number'])

column = numeric_df.columns[0]

sns.histplot(mp[column], kde=True)

plt.title(f"Skewness Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("Skewness:", mp[column].skew())
```

```python
kurt = pd.DataFrame({
    "Kurtosis": mp.kurtosis(numeric_only=True)
})

print(kurt.round(3))
```

```python
numeric_df = mp.select_dtypes(include=['number'])

column = numeric_df.columns[0]

sns.histplot(mp[column], kde=True)

plt.title(f"Kurtosis Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("Kurtosis:", mp[column].kurt())
```

```python
num = mp.select_dtypes(include=['number'])

for col in num.columns:
    gs= num[col].skew()

    if gs > 5:
        print(col, "Positive Skewed")
    elif gs < 10:
        print(col, "Negative Skewed")
    else:
        print(col, "Approximately Normal")
```
