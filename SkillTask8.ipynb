```python
tips=tips.drop_duplicates()
```

```python
tips.duplicated().sum()
```

```python
le = LabelEncoder()
tips['sex'] = le.fit_transform(tips['sex'])
tips['smoker'] = le.fit_transform(tips['smoker'])
tips['day'] = le.fit_transform(tips['day'])
tips['time'] = le.fit_transform(tips['time'])

print(tips.head())
```

```python
tips = pd.get_dummies(tips, columns=['sex', 'smoker'])

print(tips.head())
```

```python
print(tips['day'].value_counts())
```

```python
print(tips['day'].value_counts(normalize=True) * 100)
```

```python
print(tips['day'].mode())
```

```python
print(tips['time'].value_counts())
```

```python
sns.countplot(data=tips, x='day')
plt.title('Count Plot of Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

```python
tips['day'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Pie Chart")
plt.ylabel("day")
plt.show()
```

```python
sns.barplot(data=tips, x='day', y='total_bill')
plt.title("Bar Chart")
plt.show()
```
